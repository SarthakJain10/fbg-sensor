# Abstract

Fiber Bragg Grating (FBG) sensors are widely used for the measurement of strain, temperature, pressure, vibration, and other physical quantities because they offer compactness, immunity to electromagnetic interference, lightweight construction, and the possibility of multiplexing several sensing elements along a single optical fiber. The measurand is commonly encoded as a shift in the Bragg wavelength of the grating. Consequently, the performance of an FBG sensing system depends not only on the optical sensor itself but also on the interrogation technique used to detect very small wavelength shifts.

Conventional FBG interrogation methods frequently rely on detecting and tracking the spectral peak of the reflected FBG signal. The achievable wavelength resolution is therefore affected by the FBG linewidth, detector noise, spectral sampling density, and the accuracy of peak-estimation algorithms. The project considered in this report investigates an alternative interrogation architecture based on a nonlinear optical resonator and optical bistability. Instead of directly estimating the location of the FBG reflection peak, the proposed approach converts a small Bragg-wavelength displacement into a sharp switching event between two stable optical output states.

The developed numerical model represents the FBG reflection spectrum using a Lorentzian approximation and couples it to a narrower Lorentzian resonator. The resulting spectral response drives a nonlinear resonator described by a Kerr-type dispersive bistability equation. When the normalized cavity detuning exceeds the bistability threshold, the nonlinear system possesses multiple steady-state solutions and exhibits hysteresis. A wavelength displacement can consequently be detected through the occurrence of an abrupt transition between stable output states.

The project implementation consists of independent Python modules for FBG modelling, resonator modelling, bistable solving, wavelength interrogation, conventional peak tracking, resolution analysis, and visualization. The reference implementation uses a nominal Bragg wavelength of 1550 nm, an FBG linewidth of 0.2 nm, a resonator linewidth of 0.15 nm, a normalized detuning of 3, a Kerr coefficient of 1, and an input-power scale of 6 arbitrary units. The selected detuning satisfies the theoretical bistability requirement, and the corresponding fold points define the switching region of the nonlinear system.

The simulation framework generates FBG spectra, resonator responses, combined spectral responses, bistable transfer curves, hysteresis loops, noisy interrogation responses, sensitivity profiles, peak-tracking distributions, and a resolution comparison. The study establishes a numerical basis for investigating whether nonlinear bistable interrogation can provide wavelength discrimination substantially finer than conventional peak tracking. The current work is simulation-oriented; experimental validation, more complete FBG coupled-mode modelling, parameter optimization, and hardware implementation constitute important directions for future development.

**Keywords:** Fiber Bragg Grating, FBG sensor, optical fiber sensor, optical bistability, Kerr nonlinearity, nonlinear resonator, wavelength interrogation, hysteresis, spectral resolution, peak tracking.

# 1. Introduction

## 1.1 Background

Optical fiber sensing has become an important technology for applications where conventional electrical sensors may be difficult to deploy or may not provide the required immunity to electromagnetic interference. Fiber-based sensors can be made small, lightweight, electrically passive at the sensing location, and suitable for operation over long distances. Among the different fiber-optic sensing technologies, Fiber Bragg Grating (FBG) sensors have attracted considerable attention for strain, temperature, structural-health monitoring, aerospace, industrial, and environmental applications. Recent reviews identify high sensitivity and electromagnetic immunity among the principal advantages of FBG sensors, while also highlighting challenges such as temperature-strain cross-sensitivity and interrogation-system complexity.

An FBG is formed by introducing a periodic modulation of the refractive index along the core of an optical fiber. When broadband optical radiation propagates through the grating, a narrow spectral component is preferentially reflected. The central wavelength of this reflected component is known as the Bragg wavelength.

For a uniform FBG, the Bragg wavelength is approximately

$$
\lambda_B = 2n_{\mathrm{eff}}\Lambda,
$$

where

* $\lambda_B$ is the Bragg wavelength,
* $n_{\mathrm{eff}}$ is the effective refractive index of the fiber mode, and
* $\Lambda$ is the grating period.

This relationship is fundamental to FBG sensing because any physical phenomenon that changes the effective refractive index or grating period changes the Bragg wavelength.

For example, axial strain changes the grating period and also produces an elasto-optic change in the refractive index. For small strain values, the wavelength shift can be expressed approximately as

$$
\frac{\Delta\lambda_B}{\lambda_B}
=
(1-p_e)\epsilon,
$$

where $p_e$ is the effective photoelastic coefficient and $\epsilon$ is the applied strain.

Temperature affects the Bragg wavelength through both thermal expansion and the thermo-optic effect. A simplified relation is

$$
\frac{\Delta\lambda_B}{\lambda_B}
=
(\alpha+\xi)\Delta T,
$$

where $\alpha$ is the thermal expansion coefficient and $\xi$ is the thermo-optic coefficient.

Thus, an FBG can be treated as a wavelength-encoded sensor: a physical quantity changes, the Bragg wavelength changes, and an interrogation system determines the wavelength displacement.

---

## 1.2 Need for Improved FBG Interrogation

The FBG itself can possess excellent sensing characteristics, but these characteristics can only be exploited if the interrogation system can reliably detect small wavelength changes.

A conventional interrogation method is **spectral peak tracking**. The reflected spectrum is sampled, and the wavelength corresponding to the maximum reflected intensity is estimated. In an ideal noiseless system, this method is straightforward. However, real systems are affected by detector noise, finite spectral resolution, sampling limitations, spectral broadening, optical losses, and changes in the shape of the reflection spectrum.

If the wavelength shift is much smaller than the FBG linewidth, the peak of the measured spectrum may move by only a small fraction of the spectral sampling interval. The resulting estimate can become increasingly sensitive to noise.

The central question addressed in this project is therefore:

> **Can nonlinear optical processing convert a small continuous Bragg-wavelength shift into a much sharper and easier-to-detect output transition?**

The project investigates this question through a resonant-bistable interrogation architecture.

---

## 1.3 Proposed Concept

The basic concept is to place a narrowband resonator after the FBG response and operate the resonator in a nonlinear bistable regime.

The architecture can be represented conceptually as

```text
Physical measurand
       │
       ▼
┌─────────────────┐
│   FBG Sensor    │
│                 │
│ λB changes with │
│ strain/temp.    │
└────────┬────────┘
         │
         │ reflected spectrum
         ▼
┌─────────────────┐
│ Narrowband      │
│ Optical         │
│ Resonator       │
└────────┬────────┘
         │
         │ nonlinear spectral response
         ▼
┌─────────────────┐
│ Kerr Bistable   │
│ Interrogator    │
└────────┬────────┘
         │
         ▼
  Optical switching
  / output intensity
         │
         ▼
 Wavelength-shift estimate
```

The FBG is the sensing element. The nonlinear resonator is an interrogation or signal-processing element rather than the physical sensor itself. This distinction is important because the project does not propose replacing the FBG with a nonlinear resonator. Instead, it investigates nonlinear processing of the FBG's spectral output.


The repository describes the approach as a **resonant-bistable interrogation architecture** in which a narrowband FBG reflection spectrum is coupled to a nonlinear Lorentzian resonant filter. Small shifts in $\lambda_B$ can trigger switching between stable output states.

---

## 1.4 Objectives

The project aims to develop a complete numerical framework for bistable FBG interrogation. It begins at the component level, with a mathematical model of the FBG reflection spectrum and a model of a narrowband optical resonator, whose responses are then combined into a single interrogation signal. At the heart of the work is a nonlinear Kerr-type bistable resonator model: the bistable operating region and fold points are determined analytically, and wavelength sweeps through the system are simulated to demonstrate hysteresis in the optical output.
The second half of the project addresses practical performance. Realistic additive measurement noise is introduced, and the nonlinear bistable interrogation scheme is compared against conventional peak tracking, with a wavelength-resolution metric established for the proposed approach. Finally, the influence of resonator linewidth, nonlinear coefficient, input power, and noise is characterized, providing a framework for future parameter optimization and experimental implementation.

---

## 1.5 Scope of the Project

* The project is primarily focusing on simulations; no physical FBG or optical resonator is constructed experimentally.
* The work encompasses analytical modelling of the FBG and resonator, nonlinear steady-state analysis, and the calculation of bistability and fold points. These are complemented by numerical wavelength sweeps, hysteresis modelling, and additive noise analysis, together with a comparison against conventional peak tracking, resolution analysis, and graphical visualization.

The current implementation deliberately uses simplified spectral models so that the fundamental sensing concept can be studied independently of the additional complexity of a complete physical FBG structure.

# 2. Literature Survey

## 2.1 Fiber Bragg Grating Sensors

FBG sensing is based on periodic refractive-index modulation inside an optical fiber. The grating behaves as a wavelength-selective reflector, and its Bragg wavelength depends on both the effective refractive index and the grating period.

The fundamental Bragg condition is

$$
\lambda_B = 2n_{\mathrm{eff}}\Lambda.
$$

This equation also explains why FBGs can act as highly useful sensing elements. A mechanical or thermal perturbation modifies one or both quantities on the right-hand side, resulting in a measurable wavelength displacement.

FBG sensors have been investigated for structural health monitoring, aerospace systems, temperature measurement, pressure sensing, biomedical applications, and environmental monitoring. Modern reviews continue to identify the technology as useful for applications requiring high sensitivity and immunity to electromagnetic interference.

---

## 2.2 Strain Sensing

For strain sensing, the wavelength shift originates from two mechanisms:

1. mechanical modification of the grating period;
2. strain-induced modification of the refractive index.

The normalized wavelength shift is approximately

$$
\frac{\Delta\lambda_B}{\lambda_B}
=
(1-p_e)\epsilon.
$$

Consequently,

$$
\Delta\lambda_B
=
\lambda_B(1-p_e)\epsilon.
$$

This relation demonstrates why wavelength resolution is important. Even if an FBG has a nominal wavelength near 1550 nm, the wavelength shift corresponding to a small strain may be only a few picometres or less.

The literature reports typical strain-induced Bragg-wavelength sensitivities on the order of approximately $1.2$ pm/$\mu\epsilon$, although the exact value depends on fiber and packaging characteristics.

---

## 2.3 Temperature Sensing

Temperature changes affect both the grating period and the refractive index.

The general differential relationship can be written as

$$
\Delta\lambda_B
=
2
\left[
\Lambda
\frac{\partial n_{\mathrm{eff}}}{\partial T}
+
n_{\mathrm{eff}}
\frac{\partial\Lambda}{\partial T}
\right]
\Delta T.
$$

Dividing by $\lambda_B=2n_{\mathrm{eff}}\Lambda$ gives

$$
\frac{\Delta\lambda_B}{\lambda_B}
=
\left(
\frac{1}{n_{\mathrm{eff}}}
\frac{\partial n_{\mathrm{eff}}}{\partial T}
+
\frac{1}{\Lambda}
\frac{\partial\Lambda}{\partial T}
\right)
\Delta T.
$$

The two contributions correspond approximately to the thermo-optic and thermal-expansion effects. For silica, the thermo-optic contribution is generally larger than the thermal-expansion contribution.

This also illustrates an important limitation of FBG sensors: temperature and strain can both influence the measured wavelength. Therefore, multi-parameter sensing often requires compensation or multiple sensing elements.

---

## 2.4 FBG Reflection Spectrum

The actual reflection spectrum of an FBG depends on grating length, index modulation, apodization, chirp, wavelength detuning, and other physical parameters.

For the first-stage numerical investigation in this project, the FBG reflection spectrum is approximated by a Lorentzian:

$$
R_{\mathrm{FBG}}(\lambda)
=
\frac{R_0}
{
1+
\left[
\frac{2(\lambda-\lambda_B)}
{\Gamma_{\mathrm{FBG}}}
\right]^2
},
$$

where

* $R_0$ is the peak reflectivity,
* $\lambda_B$ is the Bragg wavelength, and
* $\Gamma_{\mathrm{FBG}}$ is the full width at half maximum (FWHM).

The use of a Lorentzian representation reduces the computational complexity and makes the effect of wavelength displacement easy to interpret.

The repository explicitly uses this approximation and identifies the FBG linewidth as one of the important parameters controlling the spectral response.

---

## 2.5 Conventional FBG Peak Tracking

In conventional interrogation, the reflected spectrum is measured over a wavelength interval and the peak position is estimated.

A simplified peak-tracking process is:

```text
Measure spectrum
       │
       ▼
Sample wavelength grid
       │
       ▼
Locate maximum intensity
       │
       ▼
Estimate λB
       │
       ▼
Calculate ΔλB
```

Under noise, the measured peak becomes a random variable:

$$
\hat{\lambda}_B
=
\lambda_B + e_\lambda,
$$

where $e_\lambda$ represents the estimation error.

The standard deviation

$$
\sigma_{\lambda}
=
\operatorname{std}(\hat{\lambda}_B)
$$

can be used as an empirical measure of peak-tracking resolution.

The project uses Monte Carlo simulations to generate multiple noise realizations and evaluates the resulting distribution of the estimated peak wavelength.

---

## 2.6 Nonlinear Optical Resonators

Optical resonators provide wavelength-selective transmission characteristics. A simplified Lorentzian resonator transmission is

$$
H_r(\lambda)
=
\frac{1}
{
1+
\left[
\frac{2(\lambda-\lambda_r)}
{\Gamma_r}
\right]^2
},
$$

where

* $\lambda_r$ is the resonator center wavelength;
* $\Gamma_r$ is the resonator FWHM linewidth.

A narrower $\Gamma_r$ results in a sharper resonance.

The project deliberately chooses the resonator linewidth smaller than the FBG linewidth. This creates a sharper spectral gate through which changes in the FBG wavelength are converted into stronger changes in the combined response.

---

## 2.7 Optical Bistability

Optical bistability refers to the existence of two stable output states under appropriate nonlinear operating conditions.

The project uses the normalized dispersive nonlinear resonator equation

$$
I_{\mathrm{out}}
\left[
1+
\left(
\Delta-\alpha I_{\mathrm{out}}
\right)^2
\right]
=
I_{\mathrm{in}},
$$

where

* $I_{\mathrm{out}}$ is the output or intracavity intensity,
* $I_{\mathrm{in}}$ is the effective input intensity,
* $\Delta$ is normalized cavity detuning,
* $\alpha$ is the nonlinear Kerr coefficient.

The effective input is defined as

$$
I_{\mathrm{in}}
=
I_0S(\lambda_B),
$$

where $I_0$ is the input-power scale and $S(\lambda_B)$ is the combined FBG-resonator response.

Expanding the nonlinear equation produces a cubic equation:

$$
\alpha^2I_{\mathrm{out}}^3
-
2\Delta\alpha I_{\mathrm{out}}^2
+
(1+\Delta^2)I_{\mathrm{out}}
-
I_{\mathrm{in}}
=
0.
$$

When the detuning satisfies

$$
|\Delta|>\sqrt{3},
$$

the nonlinear response can contain three real steady-state solutions over a finite input range. Two of these correspond to stable branches, while the intermediate branch is unstable.

---

## 2.8 Hysteresis and Fold Bifurcations

The transitions between the stable branches occur at fold bifurcations.

Differentiating the input-output relation gives the fold condition

$$
3\alpha^2 I_{\mathrm{out}}^2
-
4\Delta\alpha I_{\mathrm{out}}
+
(1+\Delta^2)
=
0.
$$

The fold intensities are therefore

$$
I_{\mathrm{out,fold}}
=
\frac{
2\Delta
\pm
\sqrt{4\Delta^2-3(1+\Delta^2)}
}
{3\alpha}.
$$

The existence of two real fold points creates the bistable region.

During an upward sweep, the system can remain on the lower branch until the upper switching threshold is reached. During a downward sweep, it can remain on the upper branch until the lower threshold is reached.

This produces a hysteresis loop.

The important feature for sensing is that the switching transition can occur over a very small wavelength interval. Therefore, rather than estimating the exact position of a broad spectral peak, the interrogator can detect a switching event.

---

## 2.9 Related Nonlinear Sensing Research

Recent work has investigated nonlinear optical resonators for enhancing sensing sensitivity. Nonlinear resonance can increase the slope of the optical response, thereby increasing the output response associated with a small physical perturbation.

The project represented by the supplied repository differs conceptually from approaches where the resonator itself is the sensing element. Here, the FBG remains the sensor and the nonlinear resonator acts as an interrogation stage. The project also deliberately operates inside the bistable regime and uses the switching transition as the sensing signal rather than merely exploiting increased slope immediately before bistability.

This distinction is important when positioning the work within the broader nonlinear photonics literature.

# 3. Problem Statement and Methodology

## 3.1 Problem Statement

The problem addressed by this project can be stated as follows:

> **Develop and numerically evaluate a nonlinear resonant-bistable interrogation architecture capable of detecting small FBG Bragg-wavelength shifts with potentially higher resolution than conventional spectral peak-tracking interrogation.**

The investigation is based on the hypothesis that a small shift in $\lambda_B$ can be converted into a comparatively large output-intensity transition when the FBG spectrum is coupled to a narrow resonator operating in a bistable nonlinear regime.

---

## 3.2 System Model

The complete numerical system consists of four major stages:

1. FBG spectral model;
2. resonator spectral model;
3. combined spectral response;
4. nonlinear bistable response.

The overall relationship can be summarized as

$$
\lambda_B
\longrightarrow
R_{\mathrm{FBG}}
\longrightarrow
S(\lambda_B)
\longrightarrow
I_{\mathrm{in}}
\longrightarrow
I_{\mathrm{out}}.
$$

The sensing variable is the Bragg wavelength $\lambda_B$.

---

## 3.3 FBG Model

The FBG model is

$$
R_{\mathrm{FBG}}(\lambda;\lambda_B)
=
\frac{R_0}
{
1+
\left[
\frac{2(\lambda-\lambda_B)}
{\Gamma_{\mathrm{FBG}}}
\right]^2
}.
$$

The nominal parameters used in the repository are:

| Parameter                |                  Symbol |  Value | Unit          |
| ------------------------ | ----------------------: | -----: | ------------- |
| Nominal Bragg wavelength |          $\lambda_{B0}$ | 1550.0 | nm            |
| FBG linewidth            | $\Gamma_{\mathrm{FBG}}$ |   0.20 | nm            |
| Peak reflectivity        |                   $R_0$ |    1.0 | dimensionless |

These values are explicitly specified in the project's configuration file.

---

## 3.4 Resonator Model

The resonator transmission is modelled as

$$
H_r(\lambda)
=
\frac{1}
{
1+
\left[
\frac{2(\lambda-\lambda_r)}
{\Gamma_r}
\right]^2
}.
$$

The selected parameters are

| Parameter                   |      Symbol |   Value | Unit |
| --------------------------- | ----------: | ------: | ---- |
| Resonator center wavelength | $\lambda_r$ | 1550.05 | nm   |
| Resonator linewidth         |  $\Gamma_r$ |    0.15 | nm   |

The resonator is narrower than the FBG response. The 0.05 nm offset between $\lambda_r$ and $\lambda_{B0}$ places the operating point away from the exact center of the FBG spectrum.

According to the project documentation, an offset of approximately $\Gamma_r/2$ is useful for placing the operating point on a steep part of the combined response and increasing $|dI/d\lambda_B|$.

---

## 3.5 Combined Spectral Response

The combined response is defined by

$$
S(\lambda_B)
=
R_{\mathrm{FBG}}
(\lambda_{\mathrm{probe}};\lambda_B)
H_r(\lambda_{\mathrm{probe}}),
$$

with

$$
\lambda_{\mathrm{probe}}=\lambda_r.
$$

Therefore,

$$
S(\lambda_B)
=
R_{\mathrm{FBG}}(\lambda_r;\lambda_B)
H_r(\lambda_r).
$$

Because the resonator is evaluated at its center wavelength,

$$
H_r(\lambda_r)=1,
$$

for the normalized Lorentzian model.

Consequently, the dominant dependence of $S$ on $\lambda_B$ arises from the position of the FBG reflection peak relative to the fixed resonator wavelength.

---

## 3.6 Nonlinear Bistable Model

The nonlinear response is governed by

$$
I_{\mathrm{out}}
\left[
1+
\left(
\Delta-\alpha I_{\mathrm{out}}
\right)^2
\right]
=
I_{\mathrm{in}},
$$

with

$$
I_{\mathrm{in}}
=
I_0S(\lambda_B).
$$

Expanding,

$$
\alpha^2I_{\mathrm{out}}^3
-
2\Delta\alpha I_{\mathrm{out}}^2
+
(1+\Delta^2)I_{\mathrm{out}}
-
I_{\mathrm{in}}
=
0.
$$

For a given $\lambda_B$, the simulation calculates the possible steady-state solutions of this cubic equation.

The implementation searches for the real, non-negative roots numerically and tracks the physically appropriate stable branch during wavelength sweeps.

---

## 3.7 Bistability Condition

The basic condition for bistability is

$$
|\Delta|>\sqrt{3}.
$$

The project uses

$$
\Delta=3.
$$

Since

$$
3>\sqrt{3}\approx1.732,
$$

the chosen operating point lies within the theoretically allowed bistable regime.

The nonlinear coefficient is

$$
\alpha=1.
$$

The input scale is

$$
I_0=6.
$$

These values are selected so that the sweep of the combined spectral response crosses the two fold thresholds.

For

$$

the fold intensities are obtained from

\frac{2\Delta
\pm
\sqrt{4\Delta^2-3(1+\Delta^2)}}

Substituting the parameter values gives approximately

I_{\mathrm{out,fold,1}}

and

## 3.9 Numerical Wavelength Sweep

The wavelength sweep is performed around the nominal Bragg wavelength:

$$
\lambda_B
\in
[\lambda_{B0}-0.4,\lambda_{B0}+0.4]\ \mathrm{nm}.
$$

Therefore,

$$
\lambda_B\in[1549.6,1550.4]\ \mathrm{nm}.
$$

The implementation uses 500 points for each sweep direction.

Two simulations are performed:

* upward wavelength sweep;
* downward wavelength sweep.

The upward sweep begins on the lower stable branch, while the downward sweep begins on the upper stable branch.

The resulting difference between the two paths produces the hysteresis loop.

## 3.10 Noise Model

The current simulation introduces additive Gaussian intensity noise:

$$
I_{\mathrm{measured}}
=
I_{\mathrm{out}}
+
n,
$$

where

$$
n\sim\mathcal{N}(0,\sigma_I^2).
$$

The configured noise standard deviation is

$$
\sigma_I=0.02.
$$

The project documentation describes this as approximately 0.5% relative noise for typical output intensities of approximately 3–4 arbitrary units.

Noise is important because a theoretical switching transition with infinite mathematical slope cannot be reproduced perfectly in a real detector. The practical resolution is determined by the combination of the transition slope, detector noise, sampling, and system stability.

---

## 3.11 Resolution Metric

For the bistable approach, the project defines

$$
\Delta\lambda_{\min}
=
\frac{\sigma_I}
{
\max
\left|
\frac{dI_{\mathrm{out}}}{d\lambda_B}
\right|
}.
$$

This expression follows the intuitive relationship

$$
\Delta I
\approx
\frac{dI_{\mathrm{out}}}{d\lambda_B}
\Delta\lambda_B.
$$

For a detectable intensity change comparable to the noise level $\sigma_I$,

$$
\sigma_I
\approx
\left|
\frac{dI_{\mathrm{out}}}{d\lambda_B}
\right|
\Delta\lambda_{\min}.
$$

Hence,

$$
\Delta\lambda_{\min}
\approx
\frac{\sigma_I}
{
\left|
dI_{\mathrm{out}}/d\lambda_B
\right|
}.
$$

For conventional peak tracking, the project uses Monte Carlo estimation:

$$
\Delta\lambda_{\min,\mathrm{peak}}
=
\operatorname{std}
(\hat{\lambda}_B).
$$

The improvement factor is defined as

$$
F
=
\frac{
\Delta\lambda_{\min,\mathrm{peak}}
}{
\Delta\lambda_{\min,\mathrm{bistable}}
}.
$$

A larger value of $F$ indicates greater numerical resolution improvement relative to the chosen peak-tracking baseline.

## 4.4 Development of the Nonlinear Solver

The nonlinear resonator was implemented using the cubic steady-state equation

$$
\alpha^2I_{\mathrm{out}}^3
-
2\Delta\alpha I_{\mathrm{out}}^2
+
(1+\Delta^2)I_{\mathrm{out}}
-
I_{\mathrm{in}}
=
0.
$$

The solver identifies all physically meaningful roots.

For a monostable input condition, one real positive root is normally present.

Inside the bistable region, three real roots may exist:

```text
Iout
 │
 │            Upper stable branch
 │           /
 │          /
 │    -----●
 │         /
 │        /     Unstable branch
 │       /
 │  ●---/
 │ /
 │/ Lower stable branch
 └──────────────────────── Iin
```

The lower and upper branches represent stable output states, while the middle branch represents the unstable solution.

---

## 4.5 Hysteresis Tracking

A branch-tracking mechanism was implemented to reproduce physical hysteresis.

During an upward sweep:

1. the system starts on the lower branch;
2. it follows the lower branch while stable;
3. the lower branch disappears at the fold;
4. the system jumps to the upper branch.

During a downward sweep:

1. the system starts on the upper branch;
2. it follows the upper branch;
3. the upper branch disappears at the opposite fold;
4. the system jumps to the lower branch.

This produces two different output states for the same intermediate input values, depending on the sweep history.

The repository's solver explicitly implements this branch-selection strategy.

---

## 4.6 Noise Simulation

The next stage introduced additive Gaussian intensity noise.

The noisy output is

$$
I_{\mathrm{noisy}}
=
I_{\mathrm{out}}
+
n,
\qquad
n\sim\mathcal{N}(0,0.02^2).
$$

The purpose is to determine whether the bistable transition remains distinguishable under realistic measurement uncertainty.

The repository is configured to use 2000 noise realizations for its Monte Carlo peak-tracking analysis.

---

## 4.7 Conventional Peak-Tracking Baseline

A conventional FBG interrogation method was implemented as a comparison baseline.

The process consists of:

1. generating the FBG spectrum;
2. adding measurement noise;
3. estimating the peak wavelength;
4. repeating the process over many noise realizations;
5. calculating the standard deviation of the estimated wavelength.

This creates a direct comparison between:

* **continuous spectral peak estimation**, and
* **nonlinear switching detection**.

The distinction is important because the two methods extract information from the optical signal in fundamentally different ways.

---

## 4.8 Resolution Analysis

The resolution-analysis component calculates the minimum detectable wavelength shift for the bistable approach and compares it with the Monte Carlo peak-tracking result.

The repository generates a resolution comparison figure in addition to the individual sensitivity and peak-distribution plots.

The resulting framework allows the improvement factor

$$
F=
\frac{\Delta\lambda_{\min,\mathrm{peak}}}
{\Delta\lambda_{\min,\mathrm{bistable}}}
$$

to be evaluated for different parameter combinations.

# 5. Results and Discussion

## 5.1 Verification of the Bistability Condition

The chosen normalized detuning is

$$
\Delta=3.
$$

The theoretical condition is

$$
|\Delta|>\sqrt{3}.
$$

Since

$$
3>1.732,
$$

the condition is satisfied.

This confirms that the selected nonlinear resonator can exhibit a bistable response.

The configuration file also performs a sanity check of this condition when executed.

---

## 5.2 Fold Thresholds

For

$$
\Delta=3,\qquad \alpha=1,
$$

the fold output intensities are approximately

$$
I_{\mathrm{out,1}}\approx1.184,
$$

and

$$
I_{\mathrm{out,2}}\approx2.816.
$$

The corresponding input intensities are approximately

$$
I_{\mathrm{in,low}}\approx2.911,
$$

and

$$
I_{\mathrm{in,high}}\approx5.089.
$$

With

$$
I_0=6,
$$

the corresponding normalized response levels are approximately

$$
S_{\mathrm{low}}\approx0.485
$$

and

$$
S_{\mathrm{high}}\approx0.848.
$$

Therefore, the combined spectral response must traverse this interval during the interrogation sweep to produce both switching events.

This confirms that the parameter selection is not arbitrary: the input scaling has been chosen to place the spectral response inside the bistable operating window.

---

## 5.3 Hysteresis Behaviour

One of the most important outcomes of the model is the predicted hysteresis.

When $\lambda_B$ is swept in one direction, the system follows one stable branch until the corresponding fold point is reached.

When the wavelength is swept back, the system follows a different branch and switches at the opposite threshold.

Therefore,

$$
I_{\mathrm{out}}^{\uparrow}
\neq
I_{\mathrm{out}}^{\downarrow}
$$

within the hysteresis interval.

This is useful for interrogation because a small wavelength shift can cause a large optical-output change near the switching boundary.

The effect can be interpreted as a nonlinear transformation:

$$
\text{small wavelength displacement}
\quad\longrightarrow\quad
\text{large intensity transition}.
$$

---

## 5.4 Sensitivity Enhancement Mechanism

The primary advantage of the proposed architecture comes from the slope of the output response.

For a conventional measurement,

$$
\Delta I
\approx
\frac{dI}{d\lambda_B}
\Delta\lambda_B.
$$

If

$$
\left|\frac{dI}{d\lambda_B}\right|
$$

is large, a small $\Delta\lambda_B$ produces a comparatively large $\Delta I$.

The narrow resonator helps increase the rate at which the combined spectral response changes with $\lambda_B$.

The nonlinear bistability then provides an additional switching mechanism.

The project documentation identifies three important trends:

1. reducing $\Gamma_r$ sharpens the spectral gate;
2. increasing the nonlinear coefficient can increase the switching response;
3. reducing measurement noise improves the minimum detectable wavelength shift.

---

## 5.5 Effect of Resonator Linewidth

The resonator linewidth $\Gamma_r$ is one of the most important design variables.

For a Lorentzian response,

$$
H_r(\lambda)
=
\frac{1}
{
1+
\left[
2(\lambda-\lambda_r)/\Gamma_r
\right]^2
},
$$

a smaller $\Gamma_r$ produces a steeper spectral response around resonance.

Consequently,

$$
\Gamma_r\downarrow
\quad\Rightarrow\quad
\left|
\frac{dS}{d\lambda_B}
\right|\uparrow.
$$

This generally increases the sensitivity of the interrogation architecture.

However, a narrower resonance also makes the system more sensitive to wavelength drift, fabrication tolerances, laser instability, and environmental fluctuations.

Therefore, the minimum linewidth is not automatically the optimum practical design.

---

## 5.6 Effect of Nonlinear Coefficient

The nonlinear coefficient $\alpha$ controls the strength of the Kerr interaction.

Increasing the nonlinear coefficient changes the relationship between input and output intensity and therefore modifies the position and width of the bistable operating region.

The project documentation notes that the nonlinear coefficient and input-power product determine whether the desired bistable response can be obtained.

The design problem can therefore be expressed as finding suitable values of

$$
(\alpha,\Delta,I_0)
$$

such that:

1. bistability exists;
2. the spectral sweep crosses both fold points;
3. the switching event remains robust against noise;
4. the hysteresis window is not excessively large.

---

## 5.7 Effect of Input Power

The effective input intensity is

$$
I_{\mathrm{in}}
=
I_0S.
$$

Increasing $I_0$ scales the spectral response relative to the bistable thresholds.

For the current system,

$$
I_0=6.
$$

The maximum possible drive is approximately

$$
I_{\mathrm{in,max}}
\approx6S_{\max}.
$$

The configuration was selected such that the high-response region can exceed the upper fold threshold while the low-response region can fall below the lower threshold.

Thus, $I_0$ is not simply an arbitrary optical-power parameter. It determines whether the FBG wavelength sweep can actually drive the nonlinear system through both switching points.

---

## 5.8 Effect of Detector Noise

The bistable resolution metric is directly proportional to intensity noise:

$$
\Delta\lambda_{\min}
\propto\sigma_I.
$$

Therefore,

$$
\sigma_I\downarrow
\quad\Rightarrow\quad
\Delta\lambda_{\min}\downarrow.
$$

This is intuitive: if the optical switching transition is measured with a lower-noise detector, smaller wavelength-induced output changes can be detected.

The project currently uses

$$
\sigma_I=0.02.
$$

Future experimental implementation would require consideration of:

* shot noise;
* detector electronics noise;
* laser intensity noise;
* resonator fluctuations;
* temperature drift;
* polarization fluctuations;
* wavelength drift.

---

## 5.9 Peak Tracking versus Bistable Interrogation

The two interrogation strategies can be summarized as follows:

| Characteristic              | Conventional Peak Tracking | Bistable Interrogation             |
| --------------------------- | -------------------------- | ---------------------------------- |
| Primary observable          | Peak wavelength            | Switching/output state             |
| Measurement type            | Continuous                 | Threshold/switching                |
| Main limitation             | Peak-estimation noise      | Switching stability and hysteresis |
| Dependence on FBG linewidth | Strong                     | Reduced by resonant gating         |
| Detector requirement        | Spectral measurement       | Intensity-state detection          |
| Mathematical response       | Smooth peak                | Sharp nonlinear transition         |
| Hysteresis                  | Not normally required      | Deliberately exploited             |
| Current implementation      | Monte Carlo                | Nonlinear numerical solver         |

The project therefore does not simply attempt to make the FBG peak narrower. Instead, it changes the interrogation strategy.

---

## 5.10 Dynamic Range versus Resolution

An important engineering trade-off is the distinction between **resolution** and **dynamic range**.

A very sharp switching transition may allow very small wavelength shifts to be detected. However, the bistable system only switches when the input response crosses an appropriate fold threshold.

If the wavelength displacement is too large or the operating point is incorrectly selected, the system may remain on the same branch.

Thus, the useful interrogation region is linked to the hysteresis window.

The repository explicitly notes that hysteresis width determines the dynamic range of unambiguous switching.

Therefore, optimization should not maximize sensitivity alone. It should consider an objective such as

$$
\text{Performance}
=
f(
\Delta\lambda_{\min},
\Delta\lambda_{\mathrm{dynamic}},
\sigma_I,
\text{stability}
).
$$

---

## 5.11 Numerical Nature of the Current Results

The current results should be interpreted as **numerical demonstrations**, not experimental measurements.

The project uses:

* an idealized Lorentzian FBG;
* an idealized Lorentzian resonator;
* a simplified Kerr nonlinearity;
* additive Gaussian noise;
* normalized optical quantities.

These assumptions are appropriate for studying the underlying interrogation concept, but they do not capture every physical effect present in an actual optical system.

In particular, a real FBG spectrum may contain sidelobes and non-Lorentzian features. A real resonator may have loss, coupling imperfections, thermal effects, dispersion, polarization dependence, and fabrication tolerances.

Consequently, the numerical results demonstrate feasibility of the modelling concept rather than guaranteeing a particular experimental resolution.

---

## 5.12 Limitations

The main limitations of the current implementation are:

### 5.12.1 Simplified FBG Spectrum

The FBG is represented by a Lorentzian rather than a full coupled-mode or transfer-matrix model.

### 5.12.2 Idealized Resonator

The resonator is represented by a normalized Lorentzian response without detailed physical coupling parameters.

### 5.12.3 Simplified Noise

The current model primarily uses additive Gaussian intensity noise.

### 5.12.4 No Experimental Validation

The system has not yet been validated against a physical FBG and optical resonator.

### 5.12.5 Temperature and Strain Are Not Simultaneously Modelled

The current interrogation simulation operates directly on $\lambda_B$. A complete sensor model should map strain and temperature to wavelength shift and investigate cross-sensitivity.

### 5.12.6 Parameter Optimization Is Not Yet Complete

The repository provides a framework for parameter sweeps and optimization, but systematic optimization over $(\beta,\lambda_r,\Gamma_r)$ remains future work.

# 6. Future Work

## 6.1 Full Coupled-Mode FBG Model

The first major improvement should be replacement of the Lorentzian approximation with a physical coupled-mode or transfer-matrix model.

A more complete FBG model can include:

* grating length;
* refractive-index modulation;
* apodization;
* chirp;
* sidelobes;
* nonuniform strain;
* nonuniform temperature;
* propagation loss.

The repository itself identifies a transfer-matrix or Rouard-type model as a natural extension.

This would make the simulated spectra more representative of real FBG devices.

---

## 6.2 Parameter Optimization

A systematic optimization should be performed over

$$
\theta=
(\lambda_r,\Gamma_r,\alpha,\Delta,I_0).
$$

A possible objective function is

$$
J(\theta)
=
\frac{
\Delta\lambda_{\min}(\theta)
}{
\Delta\lambda_{\mathrm{reference}}
}
+
w_1C_{\mathrm{hysteresis}}
+
w_2C_{\mathrm{instability}},
$$

where the additional terms penalize excessively narrow operating ranges or unstable configurations.

Grid search can be used initially, followed by numerical optimization using methods available in SciPy.

The repository explicitly suggests grid searches or `scipy.optimize.minimize` for this purpose.

---

## 6.3 Experimental Validation

The most important future step is experimental validation.

A possible experimental arrangement is:

```text
Broadband source
      │
      ▼
   FBG sensor
      │
      ▼
Optical filtering/
coupling stage
      │
      ▼
Nonlinear resonator
      │
      ▼
Photodetector
      │
      ▼
Data acquisition
      │
      ▼
Computer
```

The experiment should first verify the FBG spectrum and then characterize the resonator independently.

The nonlinear system should subsequently be operated at several input powers and detunings to experimentally identify the bistable window.

---

## 6.4 Temperature and Strain Calibration

The next version of the model should include explicit physical measurands.

For strain:

$$
\Delta\lambda_B
=
\lambda_B(1-p_e)\epsilon.
$$

For temperature:

$$
\Delta\lambda_B
=
\lambda_B(\alpha+\xi)\Delta T.
$$

The complete sensor model can then be written as

$$
\Delta\lambda_B
=
K_\epsilon\epsilon
+
K_T\Delta T.
$$

This would permit investigation of simultaneous strain and temperature effects.

---

## 6.5 Temperature Compensation

Because temperature and strain can both produce wavelength shifts, a single FBG does not generally distinguish the two quantities without additional information.

Future work could investigate:

* reference FBGs;
* dual-grating architectures;
* polarization-based compensation;
* cascaded FBGs;
* temperature-insensitive packaging;
* multi-parameter interrogation.

The nonlinear interrogator could potentially be extended to multiple spectral channels.

---

## 6.6 Multi-FBG Interrogation

Practical sensing networks may contain many FBG sensors on the same optical fiber.

Future work should therefore investigate whether multiple FBG peaks can be interrogated using:

* multiple resonators;
* tunable resonators;
* wavelength-multiplexed channels;
* programmable interrogation points.

A multi-channel architecture could be represented as

$$
S_i(\lambda_{B,i})
=
R_i(\lambda_r)H_{r,i}(\lambda_r),
$$

for each FBG $i$.

---

## 6.7 Improved Noise Model

A more realistic simulation should incorporate multiple noise sources:

$$
n_{\mathrm{total}}
=
n_{\mathrm{shot}}
+
n_{\mathrm{thermal}}
+
n_{\mathrm{RIN}}
+
n_{\mathrm{electronics}}.
$$

The resulting output can be expressed as

$$
I_{\mathrm{meas}}
=
I_{\mathrm{out}}
+
n_{\mathrm{total}}.
$$

Monte Carlo analysis can then be repeated using experimentally measured noise statistics.

---

## 6.8 Resonator Stability

The current approach assumes that the resonator wavelength $\lambda_r$ remains stable.

In practice,

$$
\lambda_r
=
\lambda_{r0}
+
\delta\lambda_r(t).
$$

Even small resonator drift can mimic or mask an FBG wavelength shift.

Future work should therefore investigate:

* temperature stabilization;
* reference resonators;
* active wavelength locking;
* differential interrogation;
* common-mode drift cancellation.

# 7. Conclusion

This project investigates a nonlinear resonant-bistable interrogation architecture for Fiber Bragg Grating sensors. The central idea is to transform a small Bragg-wavelength displacement into a sharp optical-output transition rather than relying exclusively on conventional spectral peak tracking.

A complete numerical framework has been developed around four principal components: an FBG reflection model, a narrowband resonator, a nonlinear Kerr bistability model, and a resolution-analysis framework. The FBG and resonator are represented initially by Lorentzian functions, while the nonlinear response is described by a cubic steady-state equation derived from the dispersive bistability relation.

For the current parameter set,

$$
\lambda_{B0}=1550\ \mathrm{nm},
$$

$$
\Gamma_{\mathrm{FBG}}=0.2\ \mathrm{nm},
$$

$$
\lambda_r=1550.05\ \mathrm{nm},
$$

$$
\Gamma_r=0.15\ \mathrm{nm},
$$

$$
\Delta=3,
$$

$$
\alpha=1,
$$

and

$$
I_0=6.
$$

The selected detuning satisfies

$$
|\Delta|>\sqrt{3},
$$

and therefore permits bistability. The analytically calculated fold input levels are approximately

$$
I_{\mathrm{in,low}}\approx2.911
$$

and

$$
I_{\mathrm{in,high}}\approx5.089.
$$

The corresponding normalized spectral-response levels are approximately 0.485 and 0.848.

The numerical framework further incorporates wavelength sweeps, branch tracking, hysteresis, additive detector noise, Monte Carlo peak tracking, and resolution analysis. The repository generates a complete collection of plots for evaluating the spectral response, bistable behavior, hysteresis, noise sensitivity, and resolution comparison.

The principal conceptual outcome is that nonlinear bistability provides a different route to FBG interrogation. Instead of attempting to estimate an extremely small displacement of a spectral maximum, the system can detect whether the nonlinear resonator has switched between stable states. The potential advantage arises from the sharpness of the nonlinear transition.

At the same time, the project demonstrates that improved resolution comes with engineering trade-offs. A narrow resonator linewidth can increase sensitivity but also increases susceptibility to drift. Stronger nonlinearity can enlarge the hysteresis region, but excessive hysteresis can reduce the usable dynamic range. Lower detector noise improves resolution, but practical systems must account for several independent noise and stability mechanisms.

The current work should therefore be regarded as a numerical proof-of-concept and modelling framework rather than a completed experimental sensor system. The most significant next steps are implementation of a physically accurate coupled-mode FBG model, systematic parameter optimization, realistic noise modelling, temperature and strain calibration, and experimental validation.

Overall, the project establishes a technically meaningful framework for investigating nonlinear optical processing as an alternative approach to high-resolution FBG interrogation.
